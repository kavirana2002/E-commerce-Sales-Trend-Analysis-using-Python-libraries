# Category Market Share (By Volume)

```python
sales_data['Product_Category'].value_counts().plot(kind = 'pie', autopct = '%1.1f%%')

plt.title("Category Distribution")
plt.show()
```

## Key Insights
### *1. Near - Equal Distribution: Electronics holds the largest slice at 21.0%, while Fashion and Home Decor are tied for the smallest ar 19.0% each.
### *2. Minimal Varience: There is only 2% difference b/w our top-performing category and our bottom- performing category.
### *3. The " Big Three": Electronic,Sports and Toys combined make up 62% of our total transaction valume.

# Customer Segment Analysis

```python
segment_sales = sales_data.groupby('Customer_Segment')['Revenue'].sum()
plt.figure()

segment_sales.plot(kind = 'pie',
                   autopct = '%1.1f%%',
                   startangle = 90,
                  explode = [0.05]*len(segment_sales))

plt.title("Revenue contribution by Customer Segment")

plt.ylabel("")
plt.tight_layout()
plt.show()
```

## Key Insights
### *1. (a)  The " Occasional" customers contribute the highest percentage of revenue at 35.2%.
###     (b)  Regular customers are right behind at 34.0%. This indicates a healthy, sustainable business model
### *2  Balanced Revenue Stream: no single segment owns more than 40% of the revenue, which protects the business from sudden drop in any one specific customer type

# Monthly Revenue Growth & Seasonality

```python
sales_data['Month'] = sales_data['Date'].dt.to_period('M')

sales_data.groupby('Month')['Revenue'].sum().plot(kind = 'line') 

plt.title("Month Revenue Trend")

plt.xlabel("Month")
plt.ylabel("Revenue")

plt.show()
```

## Key Insights
### *1.High Volatility: The graph shows revenue isn't steady; it experiences sharp peak followed by immediate drops.
### *2.Peak Performance(late 2023): Our highest revenue peak occurred just before january 2024(likely november or december 2023).
### *3.Stable baseline

# Weekly Sales Momentum

## Key Insight

# Discount Impact Analysis

```python
sales_data['Discount_Range'] = pd.cut(sales_data['Discount'],bins = [0,10,20,30,40,50])

sales_data.groupby('Discount_Range')['Units_Sold'].mean().plot(kind = 'barh')

plt.title("Discount Range vs Avg Units Sold")

plt.xlabel("Discount Range")
plt.ylabel("Avg Units Sold")

plt.show()
```

## Key Insights
### *1. Inelastic Demand: The most striking insight is that Avg Units Sold remains almost perfectly flat across all discount ranges.Whether we offer a 0-10% discount or a 40-50% discount.
### *2.Efficiency Winner: the (0,10) rangenis actually our most efficient tier.
